```python
# Setup
!pip install datasets -q
from datasets import load_dataset
import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

ds = load_dataset("FlyRank/internship-starter")
df = ds['train'].to_pandas()
print(f"Dataset loaded: {df.shape[0]:,} rows, {df.shape[1]} columns")
```

```python
# Number 1: Scale of the problem
n_clients = df['client_id'].nunique()
n_content = df['content_id'].nunique()
print(f"Number 1: {n_content:,} unique content pieces across {n_clients} clients")
print(f"          A human reviewer can check ~20 pages/week")
print(f"          ML needed to cover all {n_content:,} pages automatically")
```

```python
# Number 2: CTR opportunity exists
df['ctr_proxy'] = df['clicks_90d'] / (df['impressions_90d'] + 1)
high_imp = df[df['impressions_90d'] > df['impressions_90d'].median()]
low_ctr  = high_imp[high_imp['ctr_proxy'] < high_imp['ctr_proxy'].median()]

print(f"Number 2: {len(low_ctr):,} pages have HIGH impressions but LOW CTR")
print(f"          = {len(low_ctr)/len(df)*100:.1f}% of all content")
print(f"          These are pages already ranking — just not getting clicked")
```

```python
# Number 3: Competition level breakdown
print("Number 3: Competition level distribution:")
print(df['competition_level'].value_counts())
pct_high = (df['competition_level']=='HIGH').mean()*100
print(f"\n{pct_high:.1f}% of pages face HIGH competition")
print(f"On HIGH competition pages, even small CTR gains matter more")

# Search volume spread
print(f"\nSearch volume summary:")
print(df['search_volume'].describe(percentiles=[.50,.90]).round(1))
```
